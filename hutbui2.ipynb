```python
import random
import time

class VacuumEnvironment:
    def __init__(self, rows=4, cols=4):
        self.rows = rows
        self.cols = cols
        self.grid = [
            [random.choice(["Dirty", "Clean"]) for _ in range(cols)]
            for _ in range(rows)
        ]

    def get_status(self, position):
        row, col = position
        return self.grid[row][col]

    def clean_cell(self, position):
        row, col = position
        self.grid[row][col] = "Clean"

    def is_all_clean(self):
        return all(cell == "Clean" for row in self.grid for cell in row)

    def display(self, agent_position):
        lines = ["Environment:", ""]

        for row in range(self.rows):
            symbols = []
            for col in range(self.cols):
                if (row, col) == agent_position:
                    symbols.append(" A ")
                elif self.grid[row][col] == "Dirty":
                    symbols.append(" D ")
                else:
                    symbols.append(" C ")
            lines.append("".join(symbols))

        return "\n".join(lines)


class SimpleReflexVacuumAgent:
    def __init__(self, start_position=(0, 0)):
        self.position = start_position

    def perceive(self, environment):
        return environment.get_status(self.position)

    def build_path(self, environment):
        path = []

        for row in range(environment.rows):
            if row % 2 == 0:
                columns = range(environment.cols)
            else:
                columns = range(environment.cols - 1, -1, -1)

            for col in columns:
                path.append((row, col))

        return path

    def decide_action(self, status, environment):
        if status == "Dirty":
            return "SUCK"

        path = self.build_path(environment)
        current_index = path.index(self.position)

        if current_index == len(path) - 1:
            return "STOP"

        next_row, next_col = path[current_index + 1]
        row, col = self.position

        if next_row < row:
            return "UP"
        if next_row > row:
            return "DOWN"
        if next_col < col:
            return "LEFT"
        if next_col > col:
            return "RIGHT"

        return "STOP"

    def act(self, action, environment):
        row, col = self.position

        if action == "SUCK":
            environment.clean_cell(self.position)
        elif action == "UP":
            self.position = (row - 1, col)
        elif action == "DOWN":
            self.position = (row + 1, col)
        elif action == "LEFT":
            self.position = (row, col - 1)
        elif action == "RIGHT":
            self.position = (row, col + 1)


def print_state(step, agent, environment, status=None, action=None):
    print("-" * 45)
    print("=== SIMPLE REFLEX VACUUM CLEANER AGENT ===")
    print("A = Agent")
    print("D = Dirty")
    print("C = Clean")
    print()
    print(f"Step: {step}")
    print(f"Current position: {agent.position}")

    if status is not None:
        print(f"Current status: {status}")
    if action is not None:
        print(f"Action: {action}")

    print()
    print(environment.display(agent.position))


def run_simulation(rows=4, cols=4, delay=0):
    environment = VacuumEnvironment(rows, cols)
    agent = SimpleReflexVacuumAgent(start_position=(0, 0))
    max_steps = rows * cols * 2 + 5

    for step in range(1, max_steps + 1):
        status = agent.perceive(environment)
        action = agent.decide_action(status, environment)

        print_state(step, agent, environment, status, action)
        time.sleep(delay)

        if action == "STOP":
            break

        agent.act(action, environment)

        if environment.is_all_clean():
            print_state(step, agent, environment)
            print("\nAll cells are clean!")
            break

    if not environment.is_all_clean():
        print("\nSimulation ended. Some cells may still be dirty.")


run_simulation(rows=4, cols=4, delay=0)
```
