```python
import random

class SimpleReflexVacuumAgent:
    def __init__(self, environment):
        self.environment = environment

        # Random vị trí ban đầu của máy hút bụi
        self.location = random.choice(list(environment.keys()))

    def get_percept(self):
        # Percept là thông tin agent cảm nhận được tại vị trí hiện tại
        status = self.environment[self.location]
        return self.location, status

    def rule_action(self, percept):
        location, status = percept

        # Luật IF-THEN đơn giản
        if status == "Dirty":
            return "Suck"
        else:
            return "Move"

    def act(self):
        percept = self.get_percept()
        action = self.rule_action(percept)

        location, status = percept

        print(f"Máy đang ở ô {location}")
        print(f"Trạng thái ô hiện tại: {status}")
        print(f"Hành động: {action}")

        if action == "Suck":
            self.environment[self.location] = "Clean"
            print(f"Đã hút bụi tại ô {self.location}")

        elif action == "Move":
            # Di chuyển random sang ô khác
            possible_locations = list(self.environment.keys())
            possible_locations.remove(self.location)

            self.location = random.choice(possible_locations)
            print(f"Di chuyển sang ô {self.location}")

        print("-" * 30)


# Tạo môi trường gồm 4 ô
environment = {
    "A": random.choice(["Clean", "Dirty"]),
    "B": random.choice(["Clean", "Dirty"]),
    "C": random.choice(["Clean", "Dirty"]),
    "D": random.choice(["Clean", "Dirty"])
}

agent = SimpleReflexVacuumAgent(environment)

print("Môi trường ban đầu:")
print(environment)
print("-" * 30)

# Cho agent chạy 10 lần
for i in range(10):
    print(f"Lượt {i + 1}")
    agent.act()

print("Môi trường sau khi chạy:")
print(environment)
```

```python
# chủ đề máy hút bụi dọn sàn  0 = sạch
# 1 = dơ, lau 1 lần
# 2 = vật thể tĩnh, không đi vào
# 3 = vật có thể đụng, né
# 4 = sàn gồ ghề, phải lau 2 lần

import random

class SimpleReflexCleanerAgent:
    def __init__(self, floor):
        self.floor = floor
        self.rows = 4
        self.cols = 4
        self.position = self.random_start_position()

    def random_start_position(self):
        while True:
            r = random.randint(0, 3)
            c = random.randint(0, 3)

            # Không cho máy đứng vào vật cản
            if self.floor[r][c] != 2 and self.floor[r][c] != 3:
                return [r, c]

    def get_percept(self):
        r, c = self.position
        return self.floor[r][c]

    def clean_current_cell(self):
        r, c = self.position

        if self.floor[r][c] == 1:
            print("Ô này dơ -> lau 1 lần")
            self.floor[r][c] = 0

        elif self.floor[r][c] == 4:
            print("Ô này gồ ghề -> lau lần 1")
            print("Ô này gồ ghề -> lau lần 2")
            self.floor[r][c] = 0

    def can_move(self, r, c):
        if r < 0 or r >= self.rows or c < 0 or c >= self.cols:
            return False

        if self.floor[r][c] == 2:
            return False

        if self.floor[r][c] == 3:
            return False

        return True

    def move(self):
        directions = [
            ("lên", -1, 0),
            ("xuống", 1, 0),
            ("trái", 0, -1),
            ("phải", 0, 1)
        ]

        random.shuffle(directions)

        r, c = self.position

        for name, dr, dc in directions:
            nr = r + dr
            nc = c + dc

            if self.can_move(nr, nc):
                print(f"Di chuyển {name}: ({r}, {c}) -> ({nr}, {nc})")
                self.position = [nr, nc]
                return

        print("Không còn đường đi hợp lệ, đứng yên")

    def is_floor_clean(self):
        for row in self.floor:
            for cell in row:
                if cell == 1 or cell == 4:
                    return False
        return True

    def print_floor(self):
        for r in range(self.rows):
            for c in range(self.cols):
                if [r, c] == self.position:
                    print("A", end=" ")
                else:
                    print(self.floor[r][c], end=" ")
            print()
        print()

    def act(self):
        percept = self.get_percept()

        print(f"Máy đang ở vị trí: {self.position}")
        print(f"Percept nhận được: {percept}")

        # Tập luật IF-THEN
        if percept == 1:
            print("IF ô hiện tại dơ THEN lau")
            self.clean_current_cell()

        elif percept == 4:
            print("IF ô hiện tại gồ ghề THEN lau 2 lần")
            self.clean_current_cell()

        elif percept == 0:
            print("IF ô hiện tại sạch THEN di chuyển")
            self.move()

        print("-" * 30)


floor = [
    [0, 1, 0, 2],
    [1, 4, 0, 3],
    [0, 1, 4, 0],
    [2, 0, 1, 0]
]

agent = SimpleReflexCleanerAgent(floor)

print("Sàn ban đầu:")
agent.print_floor()

step = 1

while not agent.is_floor_clean() and step <= 50:
    print(f"Bước {step}")
    agent.act()
    agent.print_floor()
    step += 1

if agent.is_floor_clean():
    print("Đã lau sạch sàn!")
else:
    print("Chưa lau sạch hết sau 50 bước")
```
