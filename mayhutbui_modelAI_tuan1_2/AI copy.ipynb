```python
def bubble_sort(arr):
    """Bubble sort algorithm"""
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

# Test the algorithm
numbers = [64, 34, 25, 12, 22, 11, 90]
print("Original:", numbers)
print("Sorted:", bubble_sort(numbers))
```

```python
from collections import deque  # Import deque để dùng hàng đợi queue cho thuật toán BFS

class EightPuzzleAgent:
    def __init__(self, start_state):
        # Chuyển start_state từ list sang tuple
        # Vì tuple không thay đổi được nên có thể đưa vào set visited
        self.start = tuple(tuple(row) for row in start_state)

        # Trạng thái đích mà agent cần tìm tới
        # 0 đại diện cho ô trống
        self.goal = ((1,2,3),(4,5,6),(7,8,0))

        # Các hướng di chuyển của ô trống 0
        # up: giảm hàng đi 1
        # down: tăng hàng lên 1
        # left: giảm cột đi 1
        # right: tăng cột lên 1
        self.moves = {
            'up': (-1, 0),
            'down': (1, 0),
            'left': (0, -1),
            'right': (0, 1)
        }
    
    def get_neighbors(self, state):
        # Hàm này sinh ra các trạng thái tiếp theo
        # bằng cách di chuyển ô trống 0 theo các hướng hợp lệ
        neighbors = []

        # Chuyển bảng 3x3 thành list 1 chiều
        # Ví dụ:
        # ((0,2,3),(4,5,1),(6,7,8))
        # thành [0,2,3,4,5,1,6,7,8]
        flat = [tile for row in state for tile in row]

        # Tìm vị trí của ô trống 0 trong list 1 chiều
        zero_idx = flat.index(0)

        # Đổi vị trí 1 chiều thành hàng và cột trong bảng 3x3
        row, col = divmod(zero_idx, 3)

        # Duyệt qua từng hướng đi: up, down, left, right
        for move, (dr, dc) in self.moves.items():
            # Tính vị trí mới của ô trống sau khi di chuyển
            nr, nc = row + dr, col + dc

            # Kiểm tra vị trí mới có nằm trong bảng 3x3 không
            if 0 <= nr < 3 and 0 <= nc < 3:
                # Copy trạng thái hiện tại để tạo trạng thái mới
                # Không sửa trực tiếp trạng thái cũ
                new_flat = flat[:]

                # Tính index của ô sẽ đổi chỗ với ô trống
                swap_idx = nr * 3 + nc

                # Đổi chỗ ô trống 0 với ô ở vị trí mới
                new_flat[zero_idx], new_flat[swap_idx] = new_flat[swap_idx], new_flat[zero_idx]

                # Chuyển list 1 chiều trở lại bảng 3x3 dạng tuple
                new_state = tuple(tuple(new_flat[i*3:(i+1)*3]) for i in range(3))

                # Thêm trạng thái mới vào danh sách hàng xóm
                neighbors.append(new_state)

        # Trả về tất cả trạng thái có thể đi được từ state hiện tại
        return neighbors
    
    def solve(self):
        # Queue lưu từng phần tử dạng:
        # (trạng_thái_hiện_tại, đường_đi_để_tới_trạng_thái_đó)
        queue = deque([(self.start, [])])

        # visited dùng để lưu các trạng thái đã đi qua
        # giúp tránh lặp vô hạn
        visited = set([self.start])

        # Lặp cho đến khi queue rỗng
        while queue:
            # Lấy trạng thái vào queue sớm nhất ra xử lý
            # Đây là đặc điểm FIFO của BFS
            state, path = queue.popleft()

            # Nếu trạng thái hiện tại là trạng thái đích
            # thì trả về đường đi ngắn nhất
            if state == self.goal:
                return path

            # Sinh các trạng thái kế tiếp từ trạng thái hiện tại
            for neighbor in self.get_neighbors(state):
                # Nếu trạng thái này chưa từng gặp
                if neighbor not in visited:
                    # Đánh dấu là đã gặp
                    visited.add(neighbor)

                    # Đưa trạng thái mới vào queue
                    # path + [neighbor] nghĩa là:
                    # lấy đường đi cũ và thêm trạng thái mới vào cuối
                    queue.append((neighbor, path + [neighbor]))

        # Nếu duyệt hết mà không thấy goal thì trả về None
        return None


# Example usage
# Trạng thái ban đầu của bài toán
# 0 là ô trống
start = [[0,2,3],[4,5,1],[6,7,8]]

# Tạo agent với trạng thái ban đầu
agent = EightPuzzleAgent(start)

# Gọi hàm solve để tìm lời giải
solution = agent.solve()

# Nếu tìm được lời giải
if solution:
    print("Solution path:")

    # In từng trạng thái trong đường đi từ start tới goal
    for step in solution:
        for row in step:
            print(row)
        print()

# Nếu không tìm được lời giải
else:
    print("No solution found")
```

```python
from collections import deque
import random  # Dùng để tạo ma trận random

class EightPuzzleAgent:
    def __init__(self, start_state):
        self.start = tuple(tuple(row) for row in start_state)
        self.goal = ((1,2,3),(4,5,6),(7,8,0))

        self.moves = {
            'up': (-1, 0),
            'down': (1, 0),
            'left': (0, -1),
            'right': (0, 1)
        }
    
    def get_neighbors(self, state):
        neighbors = []

        flat = [tile for row in state for tile in row]
        zero_idx = flat.index(0)
        row, col = divmod(zero_idx, 3)

        for move, (dr, dc) in self.moves.items():
            nr, nc = row + dr, col + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                new_flat = flat[:]
                swap_idx = nr * 3 + nc

                new_flat[zero_idx], new_flat[swap_idx] = new_flat[swap_idx], new_flat[zero_idx]

                new_state = tuple(tuple(new_flat[i*3:(i+1)*3]) for i in range(3))
                neighbors.append(new_state)

        return neighbors
    
    def solve(self):
        queue = deque([(self.start, [])])
        visited = set([self.start])

        while queue:
            state, path = queue.popleft()

            if state == self.goal:
                return path

            for neighbor in self.get_neighbors(state):
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append((neighbor, path + [neighbor]))

        return None


# Hàm in ma trận cho đẹp
def print_matrix(state):
    for row in state:
        print(row)
    print()


# Hàm kiểm tra ma trận 8-puzzle có giải được không
def is_solvable(numbers):
    inversions = 0

    # Bỏ số 0 ra vì 0 là ô trống
    arr = [num for num in numbers if num != 0]

    # Đếm số nghịch thế
    # Nếu một số đứng trước nhưng lớn hơn số phía sau thì tính là 1 nghịch thế
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] > arr[j]:
                inversions += 1

    # Với bảng 3x3, nếu số nghịch thế chẵn thì giải được
    return inversions % 2 == 0


# Hàm tạo ma trận random giải được
def generate_random_puzzle():
    while True:
        numbers = list(range(9))  # Tạo danh sách [0,1,2,3,4,5,6,7,8]
        random.shuffle(numbers)   # Xáo trộn random

        if is_solvable(numbers):
            return [
                numbers[0:3],
                numbers[3:6],
                numbers[6:9]
            ]


# Tạo ma trận random
start = generate_random_puzzle()

# In ma trận random ban đầu
print("Ma trận random ban đầu:")
print_matrix(start)

agent = EightPuzzleAgent(start)
solution = agent.solve()

if solution:
    print("Các trạng thái để đi tới đích:")

    step_number = 1

    for step in solution:
        print(f"Bước {step_number}:")
        print_matrix(step)
        step_number += 1

else:
    print("No solution found")
```
